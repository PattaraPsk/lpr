# 01. Data Exploration & Dataset Analysis

This notebook provides exploratory data analysis (EDA) for the computer vision license plate dataset:
1. Dataset distribution across splits (train, validation, test)
2. Image dimensions and aspect ratio distributions
3. Thai character frequency and DLT consonant compliance
4. Sample visual inspection

```python
import os
from pathlib import Path
import yaml
import cv2
import numpy as np
from PIL import Image
from collections import Counter
import matplotlib.pyplot as plt

# Load dataset configuration
config_path = Path("../configs/dataset.yaml")
if not config_path.exists():
    config_path = Path("configs/dataset.yaml")

with open(config_path, "r", encoding="utf-8") as f:
    cfg = yaml.safe_load(f)

print("Dataset Configuration Loaded:")
print(yaml.dump(cfg, sort_keys=False))
```

```python
# Inspect split file statistics
splits_dir = Path("../data/splits") if Path("../data/splits").exists() else Path("data/splits")
for split_name in ["train.txt", "val.txt", "test.txt"]:
    split_file = splits_dir / split_name
    if split_file.exists():
        with open(split_file, "r", encoding="utf-8") as sf:
            lines = [line.strip() for line in sf if line.strip() and not line.startswith("#")]
        print(f"{split_name}: {len(lines)} samples")
    else:
        print(f"{split_name}: file not found")
```

```python
# Analyze Thai consonant character distribution
THAI_CONSONANTS = cfg.get("vocabulary", {}).get("thai_consonants", "กขคฆงจฉชซฌญฎฐฑฒณดตถทธนบปผฝพฟภมยรลวศษสหฬอฮ")
DIGITS = cfg.get("vocabulary", {}).get("digits", "0123456789")
EXCLUDED_OBSOLETE = ["ฃ", "ฅ", "ฏ"]

print(f"Total DLT Valid Consonants: {len(THAI_CONSONANTS)}")
print(f"Digits: {DIGITS}")
print(f"Excluded Obsolete Consonants: {EXCLUDED_OBSOLETE}")
```
